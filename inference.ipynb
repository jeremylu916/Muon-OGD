```python
from query_llm import Qwen3Generator, Qwen3ThinkGenerator

# non-thinking mode (recommended for normal QA)
gen = Qwen3Generator(
    model_name="Qwen/Qwen3-8B",   # or "Qwen/Qwen3-8B" if that is what you downloaded
    system_prompt="You are a helpful assistant."
)

prompts = [
    "What should I eat for lunch?",
    "There are three options: rice noodles, chipolet, and chinese food"
    
]

outs = gen.generate(
    prompts,
    max_new_tokens=256,
    batch_size=1,      # increase if memory allows
    temperature=0.0    # deterministic
)

for i, o in enumerate(outs, 1):
    print(f"[{i}] {o}\n")
```
